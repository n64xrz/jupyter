```python
from IPython.core.display import HTML
analytics_url = "https://analytics.prod.dart.prv.vai-compute.com/analytics/v1"
```

```python
import requests

response = requests.post(f'{analytics_url}/heatmap', json={
    "data": [
        [150, 200, 250],  # Morning traffic
        [300, 400, 350],  # Afternoon traffic
        [100, 150, 120]   # Evening traffic
    ],
    "y_axis_categories": ["Morning", "Afternoon", "Evening"],  # Time periods
    "x_axis_categories": ["Homepage", "Products", "Contact"],  # Webpage sections
    "title": "Website Traffic Heatmap",
    "xlabel": "Webpage Sections",
    "ylabel": "Time Periods"
})

if response.status_code == 200:
    print(response.json())
else:
    print(f"Failed to create heatmap. Status code: {response.status_code}")
    print(response.text)

HTML(f'<iframe src="{analytics_url}/heatmap" width="800" height="600"></iframe>')
```
